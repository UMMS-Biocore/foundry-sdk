```python
# UNIVERSAL COLLECTION MIGRATION - Handles ALL collections dynamically
import time

# Add all collections you want to migrate
collection_order = [
    "series", "patient", "patient_visit", "questionnaire_questions", "questionnaire_answers",
    "biosamp", "sample", "file", "run", "analysis", "sample_summary"
]

columns_to_remove = ["lastUpdatedUser", "owner", "perms", "__v", "createdAt", "updatedAt", "DID"]
local_canvas_id = canvasLocal["data"][0]["_id"]

def get_collection_constraints(clientLocal, local_canvas_id, collection_name):
    """Get unique and required field names for any collection"""
    try:
        time.sleep(0.2)  # Rate limiting
        collections = clientLocal.metadata.search_collections({"filter": {"canvasID": local_canvas_id, "name": collection_name}})
        if not collections.get("data"):
            return [], [], {}
        
        collection_id = collections["data"][0]["_id"]
        time.sleep(0.2)  # Rate limiting
        fields = clientLocal.metadata.get_collection_fields(collection_id)
        
        unique_fields = []
        required_fields = []
        field_mappings = {}
        
        for field in fields.get("data", []):
            field_name = field["name"]
            if field.get("unique"):
                unique_fields.append(field_name)
            if field.get("required"):
                required_fields.append(field_name)
            
            # Check if this field needs special mapping (like _id -> id)
            if field_name == "id" and field.get("required"):
                field_mappings["_id"] = "id"  # Map _id to id for required id fields
        
        return unique_fields, required_fields, field_mappings
    except Exception as e:
        print(f"    Error getting constraints for {collection_name}: {e}")
        return [], [], {}

def fix_record_for_collection(record, collection_name, canvas_id, local_canvas_id, field_mappings):
    """Fix record based on collection-specific requirements and field mappings"""
    fixed_record = record.copy()
    
    # Apply field mappings (e.g., _id -> id)
    for old_field, new_field in field_mappings.items():
        if old_field in fixed_record:
            fixed_record[new_field] = fixed_record[old_field]
            del fixed_record[old_field]
    
    # Update canvas references in all string fields
    for k, v in fixed_record.items():
        if isinstance(v, str) and v.startswith(f"{canvas_id}_"):
            fixed_record[k] = v.replace(f"{canvas_id}_", f"{local_canvas_id}_")
    
    return fixed_record

def get_existing_data_cache(clientLocal, local_canvas_id, collection_name):
    """Get existing data for conflict checking with rate limiting"""
    try:
        time.sleep(0.3)  # Rate limiting
        existing_data = clientLocal.metadata.search_data(
            canvas_id=local_canvas_id, 
            collection_name=collection_name, 
            filter_data={"take": 5000}
        )
        return existing_data.get('data', [])
    except Exception as e:
        print(f"    Warning: Could not get existing data for {collection_name}: {e}")
        return []

def check_for_conflicts(record, unique_fields, existing_records):
    """Check if record conflicts with existing data based on unique constraints"""
    conflicts = []
    for field_name in unique_fields:
        if field_name in record and record[field_name] is not None:
            field_value = record[field_name]
            for existing_record in existing_records:
                if existing_record.get(field_name) == field_value:
                    conflicts.append(f"{field_name}='{field_value}'")
                    break
    return conflicts

print("=== UNIVERSAL COLLECTION MIGRATION ===")

for canvas_item in canvas["data"]:
    canvas_id = canvas_item["_id"]
    print(f"Processing canvas: {canvas_id}")
    
    # Only keep collections whose name is in collection_order
    filtered_collections = [c for c in collections["data"] if c["name"] in collection_order]
    ordered_collections = sorted(
        filtered_collections,
        key=lambda x: collection_order.index(x["name"])
    )
    
    total_stats = {"success": 0, "conflicts": 0, "errors": 0}
    
    for collection in ordered_collections:
        collection_name = collection["name"]
        collection_id = collection["_id"]
        print(f"\n  Processing collection: {collection_name}")
        
        # Get collection constraints dynamically
        unique_fields, required_fields, field_mappings = get_collection_constraints(
            clientLocal, local_canvas_id, collection_name
        )
        print(f"    Unique fields: {unique_fields}")
        print(f"    Required fields: {required_fields}")
        if field_mappings:
            print(f"    Field mappings: {field_mappings}")
        
        # Get existing data for conflict checking
        existing_records = get_existing_data_cache(clientLocal, local_canvas_id, collection_name)
        print(f"    Found {len(existing_records)} existing local records")
        
        # Get the field order for this collection
        try:
            time.sleep(0.2)
            collection_fields = client.metadata.get_collection_fields(collection_id)
            field_order = [field["name"] for field in collection_fields.get("data", [])]
        except Exception as e:
            print(f"    Error getting field order: {e}")
            field_order = []
        
        # Get remote data
        try:
            time.sleep(0.2)
            data = client.metadata.search_data(
                canvas_id=canvas_id, 
                collection_name=collection_name, 
                filter_data={"take": 20000}
            )
            remote_records = data.get("data", [])
        except Exception as e:
            print(f"    Error getting remote data: {e}")
            continue
            
        print(f"    Found {len(remote_records)} remote records to migrate")
        
        # Clean data (remove unwanted columns)
        cleaned_data = [
            {k: v for k, v in record.items() if k not in columns_to_remove}
            for record in remote_records
        ]
        
        success_count = 0
        conflict_count = 0
        error_count = 0
        
        for i, record in enumerate(cleaned_data):
            # Apply collection-specific fixes
            fixed_record = fix_record_for_collection(
                record, collection_name, canvas_id, local_canvas_id, field_mappings
            )
            
            # Check for missing required fields
            missing_required = [f for f in required_fields 
                             if f not in fixed_record or fixed_record[f] is None or fixed_record[f] == ""]
            if missing_required:
                if i < 3:  # Only print first few errors
                    print(f"    ❌ Record {i+1} missing required fields: {missing_required}")
                error_count += 1
                continue
            
            # Check for unique constraint conflicts
            conflicts = check_for_conflicts(fixed_record, unique_fields, existing_records)
            if conflicts:
                conflict_count += 1
                if i < 3:  # Only print first few conflicts
                    print(f"    ⚠️  Record {i+1} conflicts: {conflicts} - skipping")
                continue
            
            # Reorder fields according to collection field order
            ordered_record = {}
            for field_name in field_order:
                if field_name in fixed_record:
                    ordered_record[field_name] = fixed_record[field_name]
            # Add any extra fields not in field_order
            for k, v in fixed_record.items():
                if k not in ordered_record:
                    ordered_record[k] = v
            
            try:
                response = clientLocal.metadata.create_data(
                    canvas_id=local_canvas_id,
                    collection_name=collection_name,
                    data_entry=ordered_record
                )
                success_count += 1
                
                # Update existing records cache to prevent future conflicts
                existing_records.append(ordered_record)
                
                if i < 3:  # Only print first few successes
                    record_name = ordered_record.get('name', ordered_record.get('id', f'Record {i+1}'))
                    print(f"    ✅ Record {i+1} created: {record_name}")
                    
                # Rate limiting
                if (i + 1) % 20 == 0:
                    time.sleep(0.2)
                    
            except Exception as e:
                error_count += 1
                if i < 3:
                    print(f"    ❌ Record {i+1} error: {str(e)}")
        
        collection_results = f"    Results: {success_count} success, {conflict_count} conflicts, {error_count} errors"
        print(collection_results)
        
        # Update total stats
        total_stats["success"] += success_count
        total_stats["conflicts"] += conflict_count
        total_stats["errors"] += error_count
        
        # Pause between collections to avoid rate limiting
        if success_count > 0:
            time.sleep(1)
    
    print(f"\n🎯 TOTAL MIGRATION SUMMARY:")
    print(f"   ✅ {total_stats['success']} records created successfully")
    print(f"   ⚠️  {total_stats['conflicts']} conflicts (duplicates) skipped")
    print(f"   ❌ {total_stats['errors']} errors occurred")
    print()
```

```python
# Check the migration results
print("=== MIGRATION RESULTS VERIFICATION ===")

for collection_name in collection_order:
    try:
        # Check how many records are now in each local collection
        local_data = clientLocal.metadata.search_data(
            canvas_id=local_canvas_id, 
            collection_name=collection_name, 
            filter_data={"take": 1}  # Just get count info
        )
        
        # Get the full count by checking the response structure
        search_result = clientLocal.metadata.search_data(
            canvas_id=local_canvas_id, 
            collection_name=collection_name, 
            filter_data={"take": 10000}
        )
        record_count = len(search_result.get('data', []))
        
        print(f"✅ {collection_name}: {record_count} records migrated")
        
        # Show a sample record if any exist
        if record_count > 0:
            sample_record = search_result['data'][0]
            sample_name = sample_record.get('name', sample_record.get('id', sample_record.get('_id', 'Unknown')))
            print(f"   Sample: {sample_name}")
            
        time.sleep(0.2)  # Rate limiting
        
    except Exception as e:
        print(f"❌ {collection_name}: Error checking results - {e}")

print(f"\n🎯 Migration completed for canvas: {local_canvas_id}")
print("All collections have been processed with proper field mapping, validation, and conflict detection.")
```
