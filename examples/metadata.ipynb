```python
clientLocal = ViaFoundryClient(config_path="/Users/alper/.vialocal", enable_session_history=False)
#clientLocal.configure_auth(username="alper@viascientific.com", password="nepantes", hostname="http://localhost:8081")

search_params = {
    "filter": {
        "name": "vitiligo"
    }
}
canvasLocal = clientLocal.metadata.search_canvas(search_params)
print(json.dumps(canvasLocal, indent=2))
```

```python
local_canvas_id = canvasLocal["data"][0]["_id"]

print(json.dumps(collections, indent=2))
```

```python
desired_keys = ["_id", "name", "label", "canvasID"]
selected_collection_names = {"series", "patient", "patient_visit",  "questionnaire_questions", "questionnaire_answers",
    "biosamp","sample", "file", "run", "analysis","sample_summary"}

selected_fields = [
    {k: item[k] for k in desired_keys if k in item and item[k] is not None}
    for item in collections["data"] if item["name"] in selected_collection_names
]

for item in selected_fields:
    item["canvasID"] = local_canvas_id

for item in selected_fields:
    print(json.dumps(item, indent=2))
    response = clientLocal.metadata.create_collection(item)
```

```python
del_cols=clientLocal.metadata.search_collections({"filter": {"canvasID": local_canvas_id, "name": "sample_summary"}})
print(json.dumps(del_cols, indent=2))
for item in del_cols["data"]:
    del_fields=clientLocal.metadata.search_fields({"filter": {"collectionID": item["_id"]}})
    print(json.dumps(del_fields, indent=2))
    for itemField in del_fields["data"]:
        print (f"Deleting field {itemField['name']}: {itemField['_id']}")
        clientLocal.metadata.delete_field(field_id=itemField["_id"])
    print (f"Deleting collection: {item['_id']}")
    clientLocal.metadata.delete_collection(collection_id=item["_id"])
```

```python
desired_keys = ["_id", "name", "label", "type", "deprecated", "required", "collectionID", "description", "ontology", "hidden", "unique", "identifier", "enum", "ref"]
reverse_field_collections = {"biosamp"}  # add collection names as needed

for collection in collections["data"]:
    collection_name = collection["name"]
    collection_id = collection["_id"]
    # Get fields for this collection
    collection_fields = client.metadata.get_collection_fields(collection_id)
    field_order = [field["name"] for field in collection_fields.get("data", [])]
    selected_fields = [
        {k: field[k] for k in desired_keys if k in field and field[k] is not None and field[k]}
        for field in collection_fields.get("data", [])
    ]
    # Reverse the selected_fields list for collections that need it
    if collection_name in reverse_field_collections:
        print(f"Reversing field order for collection: {collection_name}")
        selected_fields = list(reversed(selected_fields))
    ordered_items = []
    for item in selected_fields:
        # Update references in field values if needed
        for k, v in item.items():
            if isinstance(v, str) and v.startswith(f"{remote_canvas_id}_"):
                item[k] = v.replace(f"{remote_canvas_id}_", f"{local_canvas_id}_")
        # Keep the field dict order as is (since selected_fields is reversed)
        ordered_item = {k: item[k] for k in item}
        ordered_items.append(ordered_item)

    print(f"Ordered fields for collection {collection_name}: {[item['name'] for item in ordered_items]}")
    # Now create all fields after ordering
    for ordered_item in ordered_items:
        try:
            response = clientLocal.metadata.create_field(ordered_item)
            print(response)
        except Exception as e:
            if "duplicate" in str(e).lower():
                continue
            else:
                print(f"Error creating field: {e}")
```

```python
collection_order = [
    "series", "patient", "patient_visit",  "questionnaire_questions", "questionnaire_answers",
    "biosamp","sample", "file", "run", "analysis","sample_summary"
]


columns_to_remove = ["lastUpdatedUser", "owner", "perms", "__v", "createdAt", "updatedAt", "DID"]
local_canvas_id = canvasLocal["data"][0]["_id"]

for canvas_item in canvas["data"]:
    canvas_id = canvas_item["_id"]
    print(local_canvas_id)
    # Only keep collections whose name is in collection_order
    filtered_collections = [c for c in collections["data"] if c["name"] in collection_order]
    ordered_collections = sorted(
        filtered_collections,
        key=lambda x: collection_order.index(x["name"])
    )
    for collection in ordered_collections:
        collection_name = collection["name"]
        collection_id = collection["_id"]
        # Get the field order for this collection
        collection_fields = client.metadata.get_collection_fields(collection_id)
        field_order = [field["name"] for field in collection_fields.get("data", [])]
        data = client.metadata.search_data(canvas_id=canvas_id, collection_name=collection_name, filter_data={"take": 20000})
        print(f"Data for canvas {canvas_id}, collection {collection_name}:")
        cleaned_data = [
            {k: v for k, v in record.items() if k not in columns_to_remove}
            for record in data.get("data", [])
        ]
        print(json.dumps(cleaned_data, indent=2))
        for record in cleaned_data:
            # Update references in record values if needed
            for k, v in record.items():
                if isinstance(v, str) and v.startswith(f"{canvas_id}_"):
                    record[k] = v.replace(f"{canvas_id}_", f"{local_canvas_id}_")
            # Reorder fields in the record
            ordered_record = {k: record.get(k) for k in field_order if k in record}
            # Optionally, add any extra fields not in field_order at the end
            for k in record:
                if k not in ordered_record:
                    ordered_record[k] = record[k]
            try:
                response = clientLocal.metadata.create_data(
                    canvas_id=local_canvas_id,
                    collection_name=collection_name,
                    data_entry=ordered_record
                )
                print(response)
            except Exception as e:
                if "duplicate" in str(e).lower():
                    continue
                else:
                    print(f"Error creating data: {e}")
```
