```python
from viafoundry.client import ViaFoundryClient
import json
# Make sure to authenticate it
#client = ViaFoundryClient("~/.viaprod",  enable_session_history=True)
client = ViaFoundryClient(config_path="/Users/alper/.viaenv", enable_session_history=False)
#client.configure_auth(username="alper@viascientific.com", password="nepantes", hostname="http://localhost:8081")
```

```python
projects = client.call(method="POST", endpoint="/api/v1/vmeta/project/search")

print(json.dumps(projects, indent=2))
```

```python
search_params = {
    "filter": {
        "name": "vitiligo",
        "owner.name": "onur yukselen"
    }
}
project = client.metadata.search_projects(search_params)

print(json.dumps(project, indent=2))
```

```python
print(project["data"][0]["_id"])
```

```python
search_params = {
    "filter": {
        "projectID":project["data"][0]["_id"],
        "name": "patient"
    }
}
collections =  client.metadata.search_collections(search_params)
print(json.dumps(collections, indent=2))
```

```python
for collection in collections["data"]:
    search_params = {
        "filter": {
            "collectionID": collection["_id"]
        }
    }
    fields = client.metadata.search_fields(search_params)
    print(f"Fields for collection {collection['name']} ({collection['_id']}):")
    print(json.dumps(fields, indent=2))
```

```python
print(collections["data"][0]["_id"])
get_collection_fields = client.metadata.get_collection_fields(collections["data"][0]["_id"])
print(json.dumps(get_collection_fields, indent=2))
```

```python
print(project["data"][0]["_id"])
project_fields = client.metadata.get_project_fields(project["data"][0]["_id"])
print(json.dumps(get_project_fields, indent=2))
```

```python
clientLocal = ViaFoundryClient(config_path="/Users/alper/.vialocal", enable_session_history=False)
projectLocal = clientLocal.metadata.search_projects()

print(json.dumps(projectLocal, indent=2))
```

```python
new_owner = {"owner": projectLocal["data"][0]["owner"]}
print(json.dumps(new_owner, indent=2))
```

```python
project_id = projectLocal["data"][0]["_id"]
collections_new_owner = client.metadata.transfer_ownership(collections, new_owner, project_id=project_id)
print(json.dumps(collections_new_owner, indent=2))
```
