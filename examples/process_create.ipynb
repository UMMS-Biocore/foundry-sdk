```python
script_body = """
   mkdir -p normalized_data
   normalize_qc.py --input ${cleaned_data} --output normalized_data --exclude_columns ${exclude_columns}
"""
```

```python
process_config = client.process.create_process_config(
    name="new_test2",
    menu_group_name="Misc.",
    input_params=[
        {
            "name": "csv",
            "displayName": "normalized_data",
            "fileType": "csv",
            "optional": False,
            "test": "testfile.csv"
        }
    ],
    output_params=[
        {
            "displayName": "\"cleaned/*.csv\"",
            "fileType": "csv",
            "qualifier": "file",
            "optional": False
        }
    ],
    summary="",
    script_body=f"script:\n\"\"\"\n{script_body}\n\"\"\"",
    script_language="Shell",
    permission_settings={
        "viewPermissions": 3,
        "writeGroupIds": []
    },
    revision_comment="Initial revision"
)
import json
print(json.dumps(process_config, indent=2))
```

```python
client.process.create_process(process_data=process_config)
```

```python
# Update the summary of the previous process configuration
process_config['summary'] = "This process normalizes QC data and saves the output to the normalized_data directory a new change."

# Update the process on the server
client.process.update_process(process_id=129, process_data=process_config)
```
